```python
import pandas as pd
data=pd.read_csv('customer_shopping_behavior (uncleaned).csv')
```

```python
data['Review Rating']=data.groupby('Category')['Review Rating'].transform(lambda X: X.fillna(X.median))
```

```python
data.isnull().sum()
```

```python
data.columns=data.columns.str.lower()
```

```python
data.columns=data.columns.str.replace(' ','_')
```

```python
data.columns
```

```python
data=data.rename(columns={'purchase_amount_(usd)':'purchase_amount'})
```

```python
lables=['young_adult','adult','middle_aged','senior_citizen']
data['age_group']=pd.qcut (data['age'],q=4,labels=lables)
data[['age','age_group']].head()
```

```python
frequency_mapping= {
    'Fortnightly':14,
    'Weekly':7,
    'Annually':365,
    'Monthly':30,
    'Bi-Weekly':15,
    'Once':1,
    'Quarterly':90

}
data['purchase_frequency_days']=data['frequency_of_purchases'].map(frequency_mapping)
data[['purchase_frequency_days','frequency_of_purchases']].head(20)
```

```python
data=data.drop('promo_code_used',axis=1)
```
